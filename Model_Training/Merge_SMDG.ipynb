```python
BASE_PATH = "/content/drive/MyDrive/Research_Project"

DATASET_PATH = os.path.join(
    BASE_PATH,
    "Dataset"
)

OLD_CSV = os.path.join(
    BASE_PATH,
    "unified_glaucoma_dataset.csv"
)

SMDG_PATH = os.path.join(
    DATASET_PATH,
    "SMDG"
)

CSV_FILE = os.path.join(
    SMDG_PATH,
    "metadata - standardized.csv"
)

IMAGE_FOLDER = os.path.join(
    SMDG_PATH,
    "full-fundus",
    "full-fundus"
)

print("OLD CSV :", os.path.exists(OLD_CSV))
print("Metadata CSV :", os.path.exists(CSV_FILE))
print("Image Folder :", os.path.exists(IMAGE_FOLDER))
```

```python
old_df = pd.read_csv(OLD_CSV)

print(old_df.shape)

old_df.head()
```

```python
smdg = pd.read_csv(CSV_FILE)

print("Total Metadata Rows:", len(smdg))

print(smdg.columns)
```

```python
print("Unique values in types:")
print(smdg["types"].unique())

print("\nValue counts:")
print(smdg["types"].value_counts())

print("\nType expanded:")
print(smdg["type_expanded"].value_counts())
```

```python
# ============================================================
# CREATE SMDG DATAFRAME
# ============================================================

records = []

print("Reading SMDG Dataset...\n")

for _, row in smdg.iterrows():

    # Skip unknown labels
    if row["types"] == -1:
        continue

    # Image name
    image_name = os.path.basename(row["fundus"])

    # Full image path
    full_path = os.path.join(
        IMAGE_FOLDER,
        image_name
    )

    # Skip missing images
    if not os.path.exists(full_path):
        continue

    # Label
    label = int(row["types"])

    # Label name
    if label == 1:
        label_name = "Glaucoma"
    else:
        label_name = "Normal"

    records.append({

        "image_name": image_name,

        "image_path": full_path,

        "label": label,

        "label_name": label_name,

        "dataset": "SMDG"

    })

smdg_df = pd.DataFrame(records)

print("SMDG Images Added :", len(smdg_df))

print()

print(smdg_df["label_name"].value_counts())

print()

print(smdg_df.head())
```

```python
print(smdg_df.shape)

print()

print(smdg_df["label_name"].value_counts())

print()

print(smdg_df.head())
```

```python
# ============================================================
# MERGE EXISTING DATASET + SMDG
# ============================================================

final_df = pd.concat(
    [old_df, smdg_df],
    ignore_index=True
)

print("=" * 50)
print("FINAL DATASET SHAPE")
print("=" * 50)

print(final_df.shape)

print("\n")

print("=" * 50)
print("DATASET DISTRIBUTION")
print("=" * 50)

print(final_df["dataset"].value_counts())
```

```python
# ============================================================
# CLASS DISTRIBUTION
# ============================================================

print("=" * 50)
print("CLASS DISTRIBUTION")
print("=" * 50)

print(final_df["label_name"].value_counts())

print("\n")

print(final_df["label_name"].value_counts(normalize=True) * 100)
```

```python
# ============================================================
# VERIFY IMAGE PATHS
# ============================================================

missing = 0

for path in final_df["image_path"]:

    if not os.path.exists(path):
        missing += 1

print("Missing Images :", missing)
```

```python
# ============================================================
# SAVE FINAL DATASET
# ============================================================

OUTPUT_FILE = os.path.join(
    BASE_PATH,
    "unified_glaucoma_dataset_v2.csv"
)

final_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Dataset Saved Successfully")

print()

print(OUTPUT_FILE)
```
